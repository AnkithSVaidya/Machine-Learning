# PHASE C · Clean & fix

We now build the clean dataset as a **new object**, step by step, leaving `raw`
untouched. We track how many rows each step removes for the data-quality memo
(Question 7).

```python
# Start the cleaning pipeline from a copy of raw.
df = raw.copy()
removal_log = {'raw_rows': len(df)}
print('Starting rows:', len(df))
```

### 3.6 Replacing values
Standardise country labels: `'EIRE' → 'Ireland'`, `'RSA' → 'South Africa'`, and
`'Unspecified' → missing` (NaN).

```python
df['Country'] = df['Country'].replace({
    'EIRE': 'Ireland',
    'RSA': 'South Africa',
    'Unspecified': np.nan,
})
print('Country now distinct:', df['Country'].nunique())
print('Country NaN after mapping Unspecified:', df['Country'].isna().sum())
```

### 3.7 Renaming columns
Rename all columns to `snake_case`.

```python
df = df.rename(columns={
    'InvoiceNo': 'invoice_no', 'StockCode': 'stock_code', 'Description': 'description',
    'Quantity': 'quantity', 'InvoiceDate': 'invoice_date', 'UnitPrice': 'unit_price',
    'CustomerID': 'customer_id', 'Country': 'country',
})
# Attach the cancellation-initial helper from 3.4 so we can drop it in 3.11.
df['invoice_initial'] = df['invoice_no'].astype(str).str[0]
list(df.columns)
```

### 3.10 Handling missing values
Quantify missingness, then decide and justify a policy for missing `customer_id`
and blank `description`. We demonstrate `isnull()`, `fillna()`, and `dropna()`.

```python
# Quantify
print(df[['customer_id', 'description', 'country']].isnull().sum())
```

**Policy & justification**

- **Blank `description` (~1,454 rows):** these lines almost always also carry a
  `unit_price` of 0 and a non-product code — they are adjustment / bad-data lines,
  not real sales. We **drop** them (via `dropna`) when building the sales table.
- **Missing `customer_id` (~25%):** these are still *genuine completed sales*
  (real products, positive quantity and price). Dropping a quarter of the data
  would understate total revenue, best-sellers, and market size. **Decision:**
  **keep** these rows in the clean sales dataset, and tag the missing id with the
  sentinel `'GUEST'` via `fillna` so it is explicit. Customer-level questions
  (Q3 distinct customers, Q4 concentration) then simply exclude `'GUEST'`. This
  keeps revenue totals honest while not inventing customer identities.

```python
# Demonstrate fillna for the documented policy (applied to a NEW column).
df['customer_id'] = df['customer_id'].fillna('GUEST')
# For non-GUEST ids, present them as clean integer-like strings (no trailing .0)
def fmt_cid(x):
    return 'GUEST' if x == 'GUEST' else str(int(float(x)))
df['customer_id'] = df['customer_id'].apply(fmt_cid)
df['customer_id'].value_counts(dropna=False).head(5)
```

### 3.11 Deleting a column
Drop a column we will not use in the final analysis, and explain why.

```python
# `invoice_initial` was a helper for flagging cancellations (3.4 / used below).
# It is fully derivable from invoice_no and adds nothing to the final analysis,
# so we drop it to keep the schema tidy.
df = df.drop(columns=['invoice_initial'])
print('Columns after drop:', list(df.columns))
```

### 3.12 Deleting a row
Remove cancellations, non-product lines, and impossible prices/quantities to
build a clean **completed-sales** DataFrame. Each filter is counted.

```python
before = len(df)

# 1) Cancellations (InvoiceNo starts with 'C')
mask_cancel = df['invoice_no'].astype(str).str.startswith('C')
removal_log['cancellations'] = int(mask_cancel.sum())
df = df.loc[~mask_cancel]

# 2) Non-product lines (stock code is not a 5-digit product code)
mask_nonprod = ~df['stock_code'].astype(str).str.match(r'^\d{5}')
removal_log['non_product_lines'] = int(mask_nonprod.sum())
df = df.loc[~mask_nonprod]

# 3) Impossible quantities (<= 0 — returns/adjustments not already removed)
mask_qty = df['quantity'] <= 0
removal_log['nonpositive_quantity'] = int(mask_qty.sum())
df = df.loc[~mask_qty]

# 4) Impossible prices (<= 0)
mask_price = df['unit_price'] <= 0
removal_log['nonpositive_price'] = int(mask_price.sum())
df = df.loc[~mask_price]

# 5) Blank descriptions (per 3.10 policy)
mask_desc = df['description'].isnull()
removal_log['blank_description'] = int(mask_desc.sum())
df = df.dropna(subset=['description'])

print('Rows removed in Phase C row-filters:', before - len(df))
print('Rows remaining                     :', len(df))
removal_log
```

### 3.13 Dropping duplicates
Detect exact duplicates with `duplicated().sum()` and remove them with
`drop_duplicates()`. Report how many rows were dropped.

```python
dups = int(df.duplicated().sum())
print('Exact duplicate rows detected:', dups)
df = df.drop_duplicates()
removal_log['exact_duplicates'] = dups
print('Rows after de-duplication:', len(df))
```

### 3.18 Applying a function
Add a `revenue = quantity * unit_price` column, use `apply` to clean

### 3.17 Looping over a column
Show a `for`-loop / list comprehension over a column, then explain why `apply`
(vectorised C-level iteration) is the preferred, faster approach.

```python
import time

sample = df['unit_price'].head(20000)

# Explicit Python for-loop
t0 = time.perf_counter()
out_loop = []
for p in sample:
    out_loop.append(round(p * 1.20, 2))   # add 20% VAT
t_loop = time.perf_counter() - t0

# List comprehension
out_lc = [round(p * 1.20, 2) for p in sample]

# Vectorised (pandas/numpy)
t0 = time.perf_counter()
out_vec = (sample * 1.20).round(2)
t_vec = time.perf_counter() - t0

print(f'for-loop : {t_loop*1000:7.2f} ms')
print(f'vectorised: {t_vec*1000:7.2f} ms')
print(f'speed-up : ~{t_loop / max(t_vec, 1e-9):.0f}x')
```

**Why `apply`/vectorisation wins:** the explicit `for`-loop runs the Python
interpreter once per element and appends to a list; the vectorised version pushes
the whole column into compiled C/NumPy in one call. On 20k rows it is already
dozens of times faster, and the gap grows with data size — so we use vectorised
operations (and `apply` where a custom function is unavoidable) throughout.

### 3.16 Aggregating
Use `agg` to produce several statistics at once (total revenue, mean order value,
transaction count) per country and per customer.

```python
country_stats = df.groupby('country').agg(
    total_revenue=('revenue', 'sum'),
    mean_line_value=('revenue', 'mean'),
    n_lines=('revenue', 'count'),
    n_orders=('invoice_no', 'nunique'),
).sort_values('total_revenue', ascending=False)
country_stats.head(10)
```

### 3.19 Applying to groups
Use `groupby(...).apply(...)` for a custom per-customer summary — total spend and
number of active months.

```python
# Parse dates first (also needed in 3.15). Work on a copy to keep df schema clean.
df['invoice_dt'] = pd.to_datetime(df['invoice_date'], format='%m/%d/%Y %H:%M')

def customer_summary(g):
    return pd.Series({
        'total_spend': g['revenue'].sum(),
        'n_orders': g['invoice_no'].nunique(),
        'active_months': g['invoice_dt'].dt.to_period('M').nunique(),
    })

per_customer = (
    df.loc[df['customer_id'] != 'GUEST']
      .groupby('customer_id')
      .apply(customer_summary)
      .sort_values('total_spend', ascending=False)
)
per_customer.head(10)
```

Split the clean data into two periods (first vs second half of the year) and

### 3.21 Merging
Use `merge` to add a `region` column from the lookup. Compare **inner** vs
**left** joins and note which countries fail to match.

```python
left_join  = df.merge(region_lookup, on='country', how='left')
inner_join = df.merge(region_lookup, on='country', how='inner')

print('Rows  — left join :', len(left_join))
print('Rows  — inner join:', len(inner_join))
print('Rows lost by inner join (unmatched countries):', len(left_join) - len(inner_join))

unmatched = sorted(left_join.loc[left_join['region'].isna(), 'country'].dropna().unique())
print('\nCountries with no region match:', unmatched if unmatched else 'none')

# Use the left join as the canonical enriched dataset (keeps all sales rows).
sales = left_join.copy()
sales['region'] = sales['region'].fillna('Other')
sales['region'].value_counts()
```

**Note:** the **left** join keeps every sales row even when a country has no
region entry, filling the gap with `'Other'`; the **inner** join silently drops
those rows. For revenue totals we prefer the left join so no sales are lost.
`sales` is now our final, enriched, completed-sales dataset.

### Q1. Seasonality — total revenue and monthly breakdown

```python
total_revenue = sales['revenue'].sum()
print(f'Total revenue for the year: £{total_revenue:,.0f}')

monthly = sales.set_index('invoice_dt')['revenue'].resample('ME').sum()
peak_month = monthly.idxmax()
avg_other  = monthly.drop(peak_month).mean()
print(f'Peak month: {peak_month:%B %Y}  (£{monthly.max():,.0f})')
print(f'Average of other months: £{avg_other:,.0f}')
print(f'Peak lift vs typical month: +{(monthly.max()/avg_other - 1)*100:.0f}%')
```

**Interpretation:** Total cleaned revenue is about **£10.2M**. Trading climbs
steadily through the autumn and peaks in **November 2011 (~£1.45M)** — the
pre-Christmas wholesale restocking window — running roughly **double** a typical
month (≈+98%). December 2011 looks low only because the export stops on 9 Dec.
The business is strongly seasonal, so stock and cash planning should anticipate a
Q4 surge.

### Q2. Best sellers — top 10 by revenue vs top 10 by units

```python
top_rev = sales.groupby('description')['revenue'].sum().sort_values(ascending=False).head(10)
top_units = sales.groupby('description')['quantity'].sum().sort_values(ascending=False).head(10)
print('TOP 10 BY REVENUE\n', top_rev.round(0).astype('int64').to_string())
print('\nTOP 10 BY UNITS\n', top_units.to_string())
overlap = set(top_rev.index) & set(top_units.index)
print('\nProducts in BOTH lists:', len(overlap))
```

**Interpretation:** The two lists overlap only partly. High-**unit** sellers are
cheap, high-volume items (small decorations, party goods sold in bulk), while the
top-**revenue** list also includes higher-priced items where fewer units still
earn more. The difference shows revenue is driven by a mix of *volume* lines and
*value* lines — pricing and bundling strategy should treat them differently.

### Q6. Returns & cancellations — how common, by count and value

```python
# Returns/cancellations live in the RAW data (we removed them when cleaning).
raw_cancel = raw.loc[raw['InvoiceNo'].astype(str).str.startswith('C')].copy()
raw_cancel['line_value'] = raw_cancel['Quantity'] * raw_cancel['UnitPrice']

n_cancel_lines = len(raw_cancel)
cancel_value = raw_cancel['line_value'].sum()
gross_sales = (raw.loc[~raw['InvoiceNo'].astype(str).str.startswith('C')]
                  .eval('Quantity * UnitPrice')).clip(lower=0).sum()

print(f'Cancellation/return lines: {n_cancel_lines:,} '
      f'({n_cancel_lines/len(raw)*100:.1f}% of all raw lines)')
print(f'Value of returns/cancellations: £{cancel_value:,.0f}')
print(f'  ... equivalent to {abs(cancel_value)/gross_sales*100:.1f}% of gross sales value')

print('\nTOP PRODUCTS BY RETURN VALUE')
print(raw_cancel.groupby('Description')['line_value'].sum().sort_values().head(10).round(0).to_string())
print('\nTOP CUSTOMERS BY RETURN VALUE')
print(raw_cancel.dropna(subset=['CustomerID']).groupby('CustomerID')['line_value']
      .sum().sort_values().head(10).round(0).to_string())
```

**Interpretation:** Cancellations/returns are a small share of line count but a
non-trivial share of value, concentrated in a handful of products and a few large
wholesale accounts (big bulk orders that were reversed). Worth monitoring those
accounts, but it does not threaten the headline revenue picture.

### Q7. Data-quality memo — what was removed/repaired, and would we trust it?

```python
removed_total = (removal_log['cancellations'] + removal_log['non_product_lines'] +
                 removal_log['nonpositive_quantity'] + removal_log['nonpositive_price'] +
                 removal_log['blank_description'] + removal_log['exact_duplicates'])
print('--- Cleaning ledger (rows) ---')
for k, v in removal_log.items():
    print(f'{k:>24}: {v:>8,}')
print(f'{"TOTAL REMOVED":>24}: {removed_total:>8,}')
print(f'{"FINAL CLEAN ROWS":>24}: {len(sales):>8,}')
print(f'\nShare of raw rows removed: {removed_total/len(raw)*100:.1f}%')
print(f'Rows REPAIRED (missing customer_id -> GUEST): '
      f"{(sales['customer_id'] == 'GUEST').sum():,} "
      f"({(sales['customer_id'] == 'GUEST').mean()*100:.1f}% of clean rows)")
```

**Data-quality memo.** We removed roughly **one in five raw lines** —
cancellations, service/non-product codes (postage, bank charges, Amazon fees,
manual adjustments), non-positive quantities and prices, blank descriptions, and
exact duplicate lines. We **repaired** missing customer IDs by tagging them
`'GUEST'` rather than discarding ~25% of genuine sales.

**Assumptions:** (1) a 5-digit code identifies a real product; (2) negative or
zero quantity/price lines are not valid sales; (3) `'GUEST'` rows are real revenue
but cannot be attributed to a named customer.

**Would I trust it for a board report?** **Yes, for revenue, product, market, and
seasonality questions** — those rest on the cleaned, completed-sales lines and are
robust. **With one caveat:** customer-level metrics (Q3 distinct customers, Q4
concentration) cover only the ~75% of sales with a known ID, so treat them as
directional for the identified-customer base rather than the whole book.
